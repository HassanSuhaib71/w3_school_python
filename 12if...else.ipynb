## Elif

```python
a = 33
b = 33

if b > a:
   print("'b' is greater than 'a'")
elif b == a:
   print("'b' is equal to 'a'")
```

```python
score = 75

if score >= 90:
  print("Grade: A")
elif score >= 80:
  print("Grade: B")
elif score >= 70:
  print("Grade: C")
elif score >= 60:
  print("Grade: D")
```

## Else

```python
a = 200
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

## Else Without Elif

```python
a = 200
b = 33
if b > a:
  print("b is greater than a")
else:
  print("b is not greater than a")
```

```python
number = 7

if number % 2 == 0:
  print("The number is even")
else:
  print("The number is odd")
```

```python
temperature = 22

if temperature > 30:
  print("It's hot outside!")
elif temperature > 20:
  print("It's warm outside")
elif temperature > 10:
  print("It's cool outside")
else:
  print("It's cold outside!")
```

```python
username = "Emil"

if len(username) > 0:
    print(f"Welcome, {username}!")
else:
    print("Error, username cannot be empty!")
```

## Logical Operator

### AND

```python
a = 200
b = 300
c = 500

if a < b and b < c:
    print("Both are condition are ture.")
```

### OR

```python
a = 200
b = 33
c = 500
if a > b or a > c:
    print("At least one condition is true.")
```

### NOT

```python
a = 33
b = 200
if not a > b:
    print("a is not greater than b")
```

```python
age = 25
is_student = False
has_discount_code = True

if (age < 18 or age > 65) and not is_student or has_discount_code:
    print("Discount applies")
```

```python
temperature = 25
is_raining = False
is_weekend = True

if (temperature > 20 and not is_raining) or is_weekend:
    print ("Great day for outdoor activities!")
```

```python
username = "Tobias"
password = "secret123"
is_verified = True

if username and password and is_verified:
    print("Login Successful!")
else:
    print("Login failed!")
```

```python
score = 85

if score >= 0 and score <= 100:
  print("Valid score")
else:
  print("Invalid score")
```

## Nested If

```python
x = 41

if x > 10:
    print("x is grater than 10!")
    if x > 20:
        print("also greater than 20")
    else:
        print("but not above 20")
```

```python
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("You can Drive")
    else:
        ("You need license.")
else:
    print("You are underage.")
```

```python
score = 85
attendance = 90
submitted = True

if score >= 60:
    if attendance >= 80:
        if submitted:
            print("Pass Oustanding!")
        else:
            print("Pass but missing assignment!")
    else:
        print("Pass Good!")
else:
    print("Fail!")
```

### Nested If vs Logical Operators

```python
temperature = 25
is_sunny = True

if temperature > 20:
    if is_sunny:
        print("Perfect beach Weather!")

################# OR ####################

temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
  print("Perfect beach weather!")
```

```python
username = "Emil"
password = "python123"
is_active = True

if username:
  if password:
    if is_active:
      print("Login successful")
    else:
      print("Account is not active")
  else:
    print("Password required")
else:
  print("Username required")
```

```python
score = 92
extra_credit = 5

if score >= 90:
    if extra_credit > 0:
        print("A+")
    else:
        print("A")
elif score >= 80:
    print("B")
else:
    print("C")
```
