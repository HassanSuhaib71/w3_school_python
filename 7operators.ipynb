#  Ternary Operator

```python
num = 6

x = "Weekend!" if num > 5 else "Workdays!"

print(x)
```

```python
num = 6

x = "Fri" if num == 5 else "Sat" if num == 6 else "Sun" if num == 7 else "Weekdays"

print(x)
```

# Comparison Operators

```python
x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
x = 5

print(1 < x < 10)   # Chaining Comparison Operators

print(1 < x and x < 10)
```

# Logical Operators

```python
x = 5

print(x > 0 and x < 10)
```

```python
x = 5

print(x < 5 or x > 10)
```

```python
x = 5

print(not(x < 5 or x > 10))
```

# Identity Operators

```python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(z is x)
print(x == y)

print(x is not y)
```

# Membership Operators


```python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
```

```python
fruits = ["apple", "banana", "cherry"]

print("pineapple" not in fruits)
```

```python
text = "Hello World"

print("H" in text)
print("hello" in text)
print("z" not in text)
```

# Bitwise Operators

```python
print(6 & 3) # Then the & operator compares the bits and returns 0010, which is 2 in decimal.
```

```python
print(6 | 3) # Then the | operator compares the bits and returns 0111, which is 7 in decimal.
```

```python
print(6 ^ 3) # Then the ^ operator compares the bits and returns 0101, which is 5 in decimal.
```
