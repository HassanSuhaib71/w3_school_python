```python
i = 1
while i < 6:
    print(i)
    i += 1
```

### The break Statement

```python
i = 1
while i < 5:
    print(i)
    if i == 3:
        break
    i += 1
```

```python
i = 1
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

### The else Statement

```python
i = 1
while i < 6:
    print(i)
    i += 1
else:
    print("i is no longer less than 6")
```
